```python
def predict_score(study_hours, attendance):
    student = pd.DataFrame({
        "Study_Hours": [study_hours],
        "Attendance": [attendance]
    })

    prediction = model.predict(student)

    return prediction[0]
```

```python
import gradio as gr

def predict_student_score(study_hours, attendance):
    score = predict_score(study_hours, attendance)
    return round(score, 2)

demo = gr.Interface(
    fn=predict_student_score,
    inputs=[
        gr.Number(label="Study Hours"),
        gr.Number(label="Attendance (%)")
    ],
    outputs=gr.Number(label="Predicted Average Score"),
    title="Student Performance Predictor",
    description="Predict a student's average score using study hours and attendance."
)

demo.launch()
```

```python
def generate_explanation(study_hours, attendance, predicted_score):
    if predicted_score >= 85:
        level = "excellent"
    elif predicted_score >= 70:
        level = "good"
    elif predicted_score >= 50:
        level = "average"
    else:
        level = "needs improvement"

    if study_hours < 3:
        study_advice = "Try increasing study time gradually."
    else:
        study_advice = "Your study time is reasonable; maintain consistent practice."

    if attendance < 75:
        attendance_advice = "Improving attendance could help."
    else:
        attendance_advice = "Your attendance is good."

    explanation = f"""
Predicted Average Score: {predicted_score:.2f}

Performance Level: {level.capitalize()}

Study Advice: {study_advice}

Attendance Advice: {attendance_advice}
"""

    return explanation
```

```python
score = predict_score(6, 90)

print(generate_explanation(6, 90, score))
```

```python
def student_prediction_app(study_hours, attendance):
    score = predict_score(study_hours, attendance)

    explanation = generate_explanation(
        study_hours,
        attendance,
        score
    )

    return round(score, 2), explanation


demo = gr.Interface(
    fn=student_prediction_app,
    inputs=[
        gr.Number(label="Study Hours"),
        gr.Number(label="Attendance (%)")
    ],
    outputs=[
        gr.Number(label="Predicted Average Score"),
        gr.Textbox(label="Performance Explanation")
    ],
    title="🎓 Student Performance Predictor",
    description="Predict academic performance using study hours and attendance."
)

demo.launch()
```

```python
## Dataset Information

The dataset contains information about 1,000 students.

### Original Features
- Gender
- Race/Ethnicity
- Parental Level of Education
- Lunch
- Test Preparation Course
- Math Score
- Reading Score
- Writing Score

### Created Features
- Average Score
- Performance Category
```

```python
## Data Analysis & Exploratory Data Analysis (EDA)

The dataset was analyzed to understand student performance patterns.

The analysis included:
- Checking dataset shape and structure
- Checking for missing values
- Descriptive statistics
- Calculating average scores
- Comparing performance by gender
- Comparing students based on test preparation
- Comparing subject-wise average scores
- Analyzing score distributions
- Studying relationships between subjects
- Correlation analysis using a heatmap
```

```python
## Machine Learning

A Machine Learning model was developed to predict the student's average score.

### Models Used
- Linear Regression — Baseline Model
- Random Forest Regressor — Improved Model

### Model Development
- Selected relevant input features
- Applied One-Hot Encoding to categorical features
- Split the data into 80% training and 20% testing sets
- Trained the models on the training data
- Generated predictions on the test data
```

```python
## Model Evaluation

The Machine Learning models were evaluated using the following metrics:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- R² Score

These metrics were used to compare the baseline Linear Regression model with the Random Forest model.
```

```python
## Feature Importance

Feature importance was analyzed using the Random Forest model.

This analysis helps identify which student attributes contributed most to the model's predictions.

The top features were visualized using a bar chart.
```

```python
## Interactive Gradio Application

An interactive web application was created using Gradio.

The application allows users to enter student information and receive:

- Predicted Average Score
- Performance Level

The application provides a simple interface for testing the trained Machine Learning model.
```

```python
## Project Results

The project successfully:
- Analyzed student performance data.
- Identified performance patterns through EDA.
- Trained and compared Machine Learning models.
- Built a Random Forest prediction model.
- Generated predicted average scores.
- Classified predicted scores into performance levels.
- Created an interactive Gradio application.
```
